TÌM HIỂU DỮ LIỆU BREAST CANCER

```python
# ==========================================
# BÀI 1 - BREAST CANCER DATASET
# ==========================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------
# 1. Đọc dữ liệu
# ------------------------------------------

df = pd.read_csv("breast-cancer.csv")

print("5 dòng đầu tiên:")
display(df.head())


# ------------------------------------------
# 2. Xem danh sách các thuộc tính
# ------------------------------------------

print("\n===== DANH SÁCH THUỘC TÍNH =====")

for i, column in enumerate(df.columns, 1):
    print(i, "-", column)


# ------------------------------------------
# 3. Xem kích thước dữ liệu và nhãn
# ------------------------------------------

print("\n===== KÍCH THƯỚC DỮ LIỆU =====")
print("Số dòng:", df.shape[0])
print("Số cột:", df.shape[1])

print("\n===== CÁC NHÃN =====")
print(df["diagnosis"].unique())

print("\n===== SỐ LƯỢNG TỪNG NHÃN =====")
print(df["diagnosis"].value_counts())


# ------------------------------------------
# 4. Lọc các mẫu có bệnh ác tính
# ------------------------------------------

benh = df[df["diagnosis"] == "M"]

print("\n===== MẪU ÁC TÍNH (M) =====")
display(benh)

print("Số mẫu ác tính:", len(benh))


# ------------------------------------------
# 5. Lọc các mẫu lành tính
# ------------------------------------------

lanh_tinh = df[df["diagnosis"] == "B"]

print("\n===== MẪU LÀNH TÍNH (B) =====")
display(lanh_tinh)

print("Số mẫu lành tính:", len(lanh_tinh))


# ------------------------------------------
# 6. Chuyển nhãn M/B thành 1/0
# ------------------------------------------

df_corr = df.copy()

df_corr["diagnosis"] = df_corr["diagnosis"].map({
    "M": 1,
    "B": 0
})


# ------------------------------------------
# 7. Tính độ tương quan
# ------------------------------------------

correlation = (
    df_corr
    .drop(columns=["id"])
    .corr()["diagnosis"]
    .drop("diagnosis")
    .sort_values(ascending=False)
)

print("\n===== ĐỘ TƯƠNG QUAN VỚI BỆNH =====")
print(correlation)


# ------------------------------------------
# 8. Vẽ biểu đồ tương quan
# ------------------------------------------

plt.figure(figsize=(12, 8))

correlation.sort_values().plot(kind="barh")

plt.title("Độ tương quan giữa các thuộc tính và bệnh")
plt.xlabel("Hệ số tương quan")
plt.ylabel("Thuộc tính")

plt.axvline(x=0, linestyle="--")

plt.tight_layout()
plt.show()


# ------------------------------------------
# 9. Vẽ Heatmap
# ------------------------------------------

plt.figure(figsize=(16, 12))

corr_matrix = df_corr.drop(columns=["id"]).corr()

sns.heatmap(
    corr_matrix,
    cmap="coolwarm",
    center=0
)

plt.title("Ma trận tương quan giữa các thuộc tính")

plt.tight_layout()
plt.show()
```
